```python
# Loading Data
file_path = "data/SDG.pdf"
loader = PyPDFLoader(file_path)
data = loader.load()
```

```python
data
```

```python
# Concatenation of pages into one string
question_gen = ""
for page in data:
    question_gen+= page.page_content
```

```python
print(question_gen)
```

```python
from langchain_text_splitters import TokenTextSplitter
splitter_ques_gen = TokenTextSplitter(
    model_name="gpt-3.5-turbo",
    chunk_size=10000,
    chunk_overlap= 200
)
```

```python
# split raw string to token chunks
chunk_ques_gen = splitter_ques_gen.split_text(question_gen)
chunk_ques_gen
```

```python
len(chunk_ques_gen)
```

```python
type(chunk_ques_gen[0])
```

```python
# Wrapping each string inside document
from langchain_core.documents import Document
document_ques_gen = [Document(page_content= t) for t in chunk_ques_gen]
document_ques_gen
```

```python
type(document_ques_gen[0])
```

```python
splitter_ans_gen = TokenTextSplitter(
    model_name = 'gpt-3.5-turbo',
    chunk_size = 1000,
    chunk_overlap = 100
)

document_answer_gen = splitter_ans_gen.split_documents(
    document_ques_gen
)
```

```python
document_answer_gen
```

```python
llm_ques_gen_pipeline = ChatGoogleGenerativeAI(model="gemini-flash-latest", temperature=0.3)
```

```python
prompt_template = """
You are an expert at creating practice questions based on study material and documentation.
Your goal is to prepare a student or researcher for their upcoming exams and tests.
You do this by asking questions about the text below:

--------
{text}
--------

Create 10 clear, direct questions about the content of the text that will prepare the student for their test.
Make sure not to lose any important information.
IMPORTANT FORMATTING RULE: Output ONLY a numbered list of questions (1. , 2. , etc.), with one question per line. Do NOT include any introductory text, section headers, categories, or explanations.

QUESTIONS:
"""
```

```python
from langchain_core.prompts import PromptTemplate

PROMPT_QUESTIONS = PromptTemplate(template=prompt_template, input_variables=['text'])
```

```python
refine_template = """
You are an expert at creating practice questions based on study material and documentation.
Your goal is to help a student or researcher prepare for an upcoming test.
We have received some practice questions to a certain extent: {existing_answer}.
We have the option to refine the existing questions or add new ones (only if necessary) with some more context below.
-------------
{text}
-------------

Given the new context, refine the original questions in English. If the context is not helpful, please provide the original questions.
IMPORTANT FORMATTING RULE: Output ONLY a numbered list of questions (1. , 2. , etc.), with one question per line. Do NOT include any introductory text, section headers, categories, or explanations.

QUESTIONS:
"""
```

```python
REFINE_PROMPT_QUESTIONS = PromptTemplate(
    input_variables=["existing_answer", "text"],
    template = refine_template
)
```

```python
from langchain_classic.chains.summarize import load_summarize_chain
# If document_ques_gen was still str it would to failure because
# load_summarize expects list of document object to extract text from and page_content field
ques_gen_chain = load_summarize_chain(llm = llm_ques_gen_pipeline,
                                      chain_type="refine",
                                      verbose = True,
                                      question_prompt=PROMPT_QUESTIONS,
                                      refine_prompt= REFINE_PROMPT_QUESTIONS)
```

```python
from google import genai

client = genai.Client()

for model in client.models.list():
  print(model.name)
```

```python
ques = ques_gen_chain.run(document_ques_gen)

print(ques)
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import FAISS
embeddings = OllamaEmbeddings(model="nomic-embed-text")

vector_store = FAISS.from_documents(document_answer_gen, embeddings)
llm_answer_gen = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite", temperature=0.3)
```

```python
ques_list = [
    line.strip()
    for line in ques.split("\n")
    if line.strip().endswith("?") and line.strip()[0].isdigit()
]
ques_list
```

```python
from langchain_classic.chains import RetrievalQA

answer_generation_chain = RetrievalQA.from_chain_type(llm=llm_answer_gen,
                                                      chain_type="stuff",
                                                      retriever=vector_store.as_retriever())
```

```python
for question in ques_list:
  print("Question: ", question)
  answer = answer_generation_chain.run(question)
  print("Answer: ", answer)
  print("------------------------------------------\n\n")

  with open("answer.txt", "a", encoding="utf-8") as f:
    f.write(f"Question: {question}\n")
    f.write(f"Answer: {answer}\n")
    f.write("------------------------------------------\n")
```
